```python
import torch
from transformers import WavLMModel
```

```python
device = "cpu"
model = WavLMModel.from_pretrained("microsoft/wavlm-base").to(device)
model.eval()
```

```python
audio_tensor = torch.randn(48000, dtype=torch.float32).to(device)

if audio_tensor.ndim == 1:
    audio_tensor = audio_tensor.unsqueeze(0)  # Shape becomes (1, 48000)

# 3. Direct forward pass
with torch.no_grad():
    outputs = model(input_values=audio_tensor, output_hidden_states=True)

# Access intermediate layer (e.g., layer 6)
layer_6_features = outputs.hidden_states[6]  # Shape: [1, seq_len, 768]
```
